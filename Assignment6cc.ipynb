```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)
fahrenheit = (9/5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)
readings_above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings_above_32)
count = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count)
print("Vectorization performs operations on entire arrays without explicit loops.")
print("Boolean indexing selects elements satisfying a given condition.")
```

```python
import numpy as np
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
total_steps = np.sum(steps)
print("Total steps recorded:", total_steps)
mean_steps = np.mean(steps)
print("Mean number of steps:", mean_steps)
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)
print("Maximum steps:", maximum_steps)
print("Minimum steps:", minimum_steps)
daily_totals = np.sum(steps, axis=0)
print("Total steps for each day:", daily_totals)
user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", position)
print("User number:", position[0] + 1)
print("Day number:", position[1] + 1)
```

```python
import numpy as np
original = np.array([1, 2, 3, 4, 5, 6])
print("Original array:", original)
subset = original[1:4]
print("\nSubset:", subset)
subset[0] = 999
print("\nAfter modifying subset:")
print("Original array:", original)
print("Subset:", subset)
copied = original.copy()
copied[0] = 500
print("\nAfter modifying copied array:")
print("Original array:", original)
print("Copied array:", copied)
arr = np.arange(1, 13).reshape(3, 4)
print("\n3 x 4 matrix:")
print(arr)
print("\nFirst row:", arr[0])
print("Last row:", arr[-1])
print("Second column:", arr[:, 1])
print("Rows 1-2 and columns 2-3:")
print(arr[1:3, 1:3])
flat1 = arr.flatten()
flat2 = arr.ravel()
print("\nUsing flatten():", flat1)
print("Using ravel():", flat2)
ravel_array = arr.ravel()
ravel_array[0] = 100
print("\nAfter modifying ravel array:")
print("Ravel array:", ravel_array)
print("Original matrix:")
print(arr)
flatten_array = arr.flatten()
flatten_array[1] = 200
print("\nAfter modifying flatten array:")
print("Flatten array:", flatten_array)
print("Original matrix:")
print(arr)
print("\nShape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
import numpy as np
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
print("(a) Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
XT = X.T
print("\n(b) Transpose of X:")
print(XT)
product = XT @ X
print("\n(c) Matrix product X.T @ X:")
print(product)
inverse = np.linalg.inv(product)
print("\n(d) Inverse of X.T @ X:")
print(inverse)
beta = inverse @ XT @ y
print("\n(e) Least-squares coefficients:")
print(beta)
print("\nThe coefficients represent the estimated contribution")
print("of sleep hours, activity level and stress level.")
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\n(g) New user features:", new_user)
print("Predicted assistance score:", predicted_score)
```
