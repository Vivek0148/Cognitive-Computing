```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
reversed_arr = arr[::-1]
print("Original Array:", arr)
print("Reversed Array:", reversed_arr)
```

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]
print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)
```

```python
import numpy as np
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3,])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]
print("Array:", y)
print("Most frequent value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)
```

```python
import numpy as np
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print("Original 2D Array:")
print(arr)
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np
Vivek_Pandey = np.linspace(10, 100, 25)
print("Original Array:")
print(Vivek_Pandey)
print("\nDimensions:", Vivek_Pandey.ndim)
print("Shape:", Vivek_Pandey.shape)
print("Total Elements:", Vivek_Pandey.size)
print("Data Type:", Vivek_Pandey.dtype)
print("Total Bytes Consumed:", Vivek_Pandey.nbytes)
reshaped_array = Vivek_Pandey.reshape(1, 25)
print("\nTranspose using reshape():")
print(reshaped_array.T)
print("\nTranspose using T attribute:")
print(Vivek_Pandey.reshape(1, 25).T)
```

```python
import numpy as np
ucs420_vivek_pandey = np.array([
    [-10, 20, 30, 50],
    [60, 70, 80, 90],
    [10, 20, 30, 50]
])
print("Original Array:")
print(ucs420_vivek_pandey)
print("\nMean:", np.mean(ucs420_vivek_pandey))
print("Median:", np.median(ucs420_vivek_pandey))
print("Maximum:", np.max(ucs420_vivek_pandey))
print("Minimum:", np.min(ucs420_vivek_pandey))
print("Unique Elements:", np.unique(ucs420_vivek_pandey))
reshaped_ucs420_vivek_pandey = ucs420_vivek_pandey.reshape(4, 3)
print("\nReshaped Array (4 rows and 3 columns):")
print(reshaped_ucs420_vivek_pandey)
resized_ucs420_vivek_pandey = np.resize(ucs420_vivek_pandey, (2, 3))
print("\nResized Array (2 rows and 3 columns):")
print(resized_ucs420_vivek_pandey)
```
