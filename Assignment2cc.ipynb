```python
import random
random.seed(1024170051)
rand_list=[random.randint(100, 900) for _ in range(100)]
print(rand_list)
```

```python
odd_numbers=[x for x in rand_list if x % 2 != 0]
print(len(odd_numbers))
print(odd_numbers)
```

```python
even_numbers=[x for x in rand_list if x % 2 == 0]
print(len(even_numbers))
print(even_numbers)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_list=[x for x in rand_list if is_prime(x)]
print(len(prime_list))
print(prime_list)
```

```python
most_common=max(set(rand_list), key=rand_list.count)
most_common_count=rand_list.count(most_common)
print(f"{most_common}, occurs {most_common_count} time(s)")
```

```python
my_dict = {
    "name": "Vivek Pandey",
    "roll_no": "1024170051",
    "branch": "COPC",
    "age": 20,
    "city": "Karnal"
}
```

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
my_dict["cgpa"] = 8.5
print(my_dict)
```

```python
my_dict["age"] += 1
print(my_dict)
```

```python
dict_copy1=my_dict.copy()
dict_copy2=my_dict.copy()
removed_value=dict_copy1.pop("branch")
print("dict_copy1:", dict_copy1)
print("pop() returned:", removed_value)
del dict_copy2["branch"]
print("dict_copy2:", dict_copy2)
```

```python
for key, value in my_dict.items():
    print(f"{key} -> {value}")
```

```python
if "email" in my_dict:
    print("Email:",my_dict["email"])
else:
    print("'email' key not found - no email on record")
```
